```python
# To Select random words we need to import random library
import random
# List containing words which should be guessed
words=["Hello","World","Python","Task","CodeAlpha"]
# Randomly select the word from the list
word=random.choice(words)
# Created a underline for each letter of the selected word
display=["_"]*len(word)
# Display the hidden word
print(" ".join(display))
# Stored the letters that are already guessed by the user
guessed_letters=[]
# Keep Track of number of wrong guesses.
wrong_guesses=0
# Continue the game until the player wins or reaches 6 wrong guesses
while wrong_guesses<6 and "_" in display:
    # Ask the player to guess the letter of hidden word.
    guess=input("Guess a letter ")
    # Check whether the letter is already guessed
    if guess in guessed_letters:
        print("You already guessed this letter")
    else:
        # If letter is not guessed store it in the list
        guessed_letters.append(guess)
        # Check if the guessed letter exists in the word.
        if guess in word:
            # Check each position of the word
            for i in range(len(word)):
                # Reveal the letter if it matches.
                if word[i]==guess:
                    display[i]=guess
            # Display the updated word
            print(" ".join(display))
        else:
            # Increase the wrong guess count
            wrong_guesses+=1
            # Calculate the remaining attempts
            rem_attempts=6-wrong_guesses
            # Display the remaining attempts
            print(rem_attempts," attempts are remaining")
# Check if all letters have been guessed
if "_" not in display:
    print("You won")
    print("Your word is ",word)
# Check if the player has used all 6 wrong guesses
elif wrong_guesses==6:
    print("You are lost. The exact word is ")
    print(word)
```
